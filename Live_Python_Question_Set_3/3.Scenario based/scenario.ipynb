```python
# 3.

def anagram(word1,word2):
    str1 = word1.lower().replace(" ","")
    str2 = word2.lower().replace(" ","")
    if sorted(str1) == sorted(str2):
        print("Anagram")
    else:
        print("Not a Anagram")
anagram("Listen","Silent")
```

```python
# 4.
def vowels(word):
    vowels = "aeiouAEIOU"
    count=0
    for char in word:
        if char in vowels:
            count += 1
    print(count)
vowels("Education")   
```

```python
# 5.
def reverse_words(sentense):
    words = sentense.split()
    reverse = words[::-1]
    reversed_sentense = " ".join(reverse)
    print(reversed_sentense)
reverse_words("Python is fun to learn")
```

```python
#6.
def missing(array_list):
    current_sum = sum(array_list)
    n = len(array_list)+1
    expected_sum = (n * (n + 1)) // 2
    return expected_sum - current_sum
missing([1,2,4,5])
```

```python
# 7
def bankingsystem():
    balance = float(input("Enter your current balance"))
    withdrawal = float(input("Enter your withdrawal amount"))

    if withdrawal <= 0:
        print("Invalid amount")
    elif withdrawal <= balance:
        balance = balance - withdrawal
        print("withdrawal successfully!")
        print("Ramaining balance:",balance)
    else:
        print("insufficient balance")
        print("Available balance is:",balance)
bankingsystem()
```

```python
#8
def find_all_duplicates(data):
    seen = set()
    duplicates = set()
    for item in data:
        if item in seen:
            duplicates.add(item)
        else:
            seen.add(item)
    return list(duplicates)
data = [1, 2, 3, 1, 4, 3, 5, 1]
find_all_duplicates(data)
```

```python
#9
def sum_of_digits(number):
    clean_number = abs(number)
    total = 0 
    
    for digit in str(clean_number):
        total += int(digit)  
        
    return total

print(sum_of_digits(1234))  
print(sum_of_digits(-56))  
print(sum_of_digits(0))     
```

```python
#10
import string
def pangram(sentense):
    set1 = set(string.ascii_lowercase)
    set2 = set(sentense.lower())
    return set1.issubset(set2)
pangram("The quick brown fox jumps over the lazy dog")
# pangram("Hello world")
```
