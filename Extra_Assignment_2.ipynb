```python
# print Odd Numbers in the list
numbers = (20, 10, 16, 19, 25, 1, 276, 188)
print(numbers)
for n in numbers:
    if n % 2 != 0:
        print(n, "is odd")
```

```python
# print Even numbers in the list
print(numbers)
for n in numbers:
    if n % 2 == 0:
        print(n, "is even")
```
