```python
#8.
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# Creating a dataset with missing values
X_train = np.array([[1, 2], [3, np.nan], [5, 6]])
y_train = np.array([10, 20, 30])
df = pd.DataFrame(X_train)
mean_median = df.mean()
df.fillna(mean_median, inplace=True)
X_train = df.to_numpy()
model = LinearRegression()
model.fit(X_train, y_train)  
print("Cleaned data:\n", X_train)
print("Predictions:", model.predict(X_train))
```

```python
#9.
from sklearn.tree import DecisionTreeClassifier

X_train = [[1, 2], [3, 4], [5, 6]]
y_train = ["spam", "ham", "spam"]  # Error: String labels not allowed

model = DecisionTreeClassifier()
model.fit(X_train, y_train)
# Predict class
print("Prediction:", model.predict([[2, 3]]))

# Predict probabilities
print("Probabilities:", model.predict_proba([[3, 4]]))

# Check available classes
print("Classes:", model.classes_)
```
