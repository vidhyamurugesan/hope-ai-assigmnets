```python
import pandas as pd

data = {'Customer_ID': [101, 102, 103, 101, 104, 102, 101, 105, 102, 103],
        'Purchase_Amount': [200, 150, 180, 220, 300, 200, 100, 400, 250, 300]}
df = pd.DataFrame(data)

# Count the number of purchases per customer
freq = df.groupby('Customer_ID').size().reset_index(name='Purchase_Count')
print("Customer Purchase Frequency:")
print(freq)

# Top 3 customers with the highest number of purchases
top3 = freq.nlargest(3, 'Purchase_Count').reset_index(drop=True)
print("\nTop 3 Frequent Customers:")
print(top3)
```

```python
import pandas as pd

data = {'Student_ID': [201, 202, 203, 201, 204, 202, 201, 205, 202, 203],
        'Exam_Score': [85, 90, 78, 88, 92, 87, 80, 95, 89, 84]}
df = pd.DataFrame(data)

# Average exam score per student
avg = df.groupby('Student_ID')['Exam_Score'].mean().round(2).reset_index(name='Avg_Score')
print("Average Exam Scores per Student:")
print(avg)

# Top 3 students with the highest average score
top3 = avg.nlargest(3, 'Avg_Score').reset_index(drop=True)
print("\nTop 3 Students:")
print(top3)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {'Size_sqft': [1500, 1800, 2400, 3000, 3500, 4000],
        'Price': [300000, 350000, 450000, 550000, 650000, 700000]}
df = pd.DataFrame(data)

independant = df[['Size_sqft']]
dependant = df['Price']

regressor = LinearRegression()
regressor.fit(independant, dependant)

# Predict the price of a 2800 sqft house
new_house = pd.DataFrame({'Size_sqft': [2800]})
predicted = regressor.predict(new_house)[0]
print(f"Predicted Price for 2800 sqft: ${int(predicted)}")
```

```python
import pandas as pd

data = {'Product_ID': ['P101', 'P102', 'P103', 'P101', 'P104', 'P102', 'P101', 'P105', 'P102', 'P103'],
        'Purchase_Count': [5, 3, 4, 2, 1, 6, 7, 3, 2, 5]}
df = pd.DataFrame(data)

# Total purchases per product
totals = df.groupby('Product_ID')['Purchase_Count'].sum().reset_index(name='Total_Purchases')
print("Product Purchase Counts:")
print(totals)

# Top 3 most purchased products
top3 = totals.nlargest(3, 'Total_Purchases').reset_index(drop=True)
print("\nTop 3 Purchased Products:")
print(top3)
```
