```python
import pandas as pd
dataset=pd.read_csv("CKD.csv")
dataset
```

```python
#pre-processing dataset by converting string to number(nominal data)
dataset=pd.get_dummies(dataset,drop_first=True,dtype=int)
dataset.columns
```

```python
# chceking the dataset is balanced or imbalanced-if both 0 and 1 has almost equal values then its balanced otherwise its imbalanced dataset
dataset['classification_yes'].value_counts()
```

```python
# spliting input and output varaibales
independant=dataset[['age', 'bp', 'al', 'su', 'bgr', 'bu', 'sc', 'sod', 'pot', 'hrmo', 'pcv',
       'wc', 'rc', 'sg_b', 'sg_c', 'sg_d', 'sg_e', 'rbc_normal', 'pc_normal',
       'pcc_present', 'ba_present', 'htn_yes', 'dm_yes', 'cad_yes',
       'appet_yes', 'pe_yes', 'ane_yes']]
dependant=dataset['classification_yes']
```

```python
# number of rows and columns in independant variable
independant.shape
```

```python
# number of rows and columns in dependant variable
dependant.shape
```

```python
#train and test data split
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(independant,dependant,test_size=0.30,random_state=0)
```

```python
#Support vector machine - support vector classification
```

```python
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import confusion_matrix,classification_report,roc_auc_score
#hyperparamter tuning
grid_params={'kernel':['rbf','poly','sigmoid','linear'], 'C': [0.1, 1, 5, 10]}
# Creating a parameter grid to experiment with different parameter values
grid=GridSearchCV(SVC(probability=True),grid_params,refit=True,verbose=3,n_jobs=-1,scoring='f1')
#model creation phase
grid.fit(X_train,Y_train)
```

```python
#model testing
y_predict=grid.predict(X_test)
#evaluation mterics
cfmatrix=confusion_matrix(Y_test,y_predict)
print(cfmatrix)
cf_report=classification_report(Y_test,y_predict)
print(cf_report)
# Calculate probabilities for the positive class (class 1)
y_probability = grid.predict_proba(X_test)[:, 1]
# print(y_probability)
# Calculate ROC AUC score
roc_score = roc_auc_score(Y_test, y_probability)
print(roc_score)
```
