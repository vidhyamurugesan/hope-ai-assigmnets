```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset=pd.read_csv("insurance_pre.csv")
dataset
```

```python
dataset=pd.get_dummies(dataset,drop_first=True,dtype=int)
dataset
```

```python
# dataset.columns
independant=dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
independant
```

```python
dependant=dataset[["charges"]]
dependant
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(independant,dependant,test_size=0.30,random_state=0)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
grid_params={'kernel':['rbf','poly','sigmoid','linear'],'C':[10,100,1000,2000,3000]}
grid = GridSearchCV(SVR(),grid_params,refit=True,verbose=3,n_jobs=-1)
grid.fit(X_train,Y_train)
```

```python
result=grid.cv_results_
result
```

```python
table=pd.DataFrame.from_dict(result)
table
```

```python
future_prrdiction=grid.predict([[32,42,2,0,1]])
future_prrdiction
```
