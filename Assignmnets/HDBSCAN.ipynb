```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv("Mall_Customers.csv")
dataset
```

```python
X=dataset.iloc[:,[3,4]].values
```

```python
from sklearn.cluster import HDBSCAN
db = HDBSCAN(min_cluster_size=7)
y_mean=db.fit_predict(X)
y_mean
```

```python
# supervised=pd.DataFrame(dataset)
dataset['Cluster_group']=y_mean
dataset
```

```python
import seaborn as sns
facet=sns.lmplot(data=dataset,x=dataset.columns[3],y=dataset.columns[4],hue=dataset.columns[5],fit_reg=False,legend=True)
```
