```python
import pandas as pd
dataset = pd.read_csv("insurance_pre.csv")
dataset
```

```python
# convert nominal data into numerical
dataset=pd.get_dummies(dataset)
```

```python
# dataset.columns
independant=dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
independant
```

```python
dependant=dataset[["charges"]]
dependant
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(independant,dependant,test_size=0.30,random_state=0)
```

```python
from sklearn.tree import DecisionTreeRegressor
regressr=DecisionTreeRegressor(criterion='squared_error',splitter='random')
regressr.fit(X_train,Y_train)
```

```python
# import matplotlib.pyplot as plt
# from sklearn import tree
# tree.plot_tree(regressr)
# plt.show()
```

```python
predict=regressr.predict(X_test)
predict
```

```python
from sklearn.metrics import r2_score
score=r2_score(Y_test,predict)
score
```

```python
import pickle
filename="finalized_multi_linear_regression.sav"
pickle.dump(regressr,open(filename,'wb'))
```
