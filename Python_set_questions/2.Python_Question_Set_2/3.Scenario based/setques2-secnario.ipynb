```python
#1. Check whether a number is positive, negative, or zero
def check_number(number):
    if number > 0:
        print("Positive")
    elif number < 0:
        print("Negative")
    else:
        print("Zero")

check_number(-10)
```

```python
# 2. Find the sum of digits of a number
def sum_of_digits(number):
    total = 0

    while number > 0:
        digit = number % 10
        total += digit
        number = number // 10

    return total

print(sum_of_digits(1234))
```

```python
def reverse_number(number):
    reverse = 0

    while number > 0:
        number, reminder = divmod(number, 10)
        reverse = reverse * 10 + reminder

    return reverse

print(reverse_number(12345))
```

```python
# Check whether a number is prime
def check_prime(number):
    if number <= 1:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False
    return True

print(check_prime(15))
```

```python
# Find factorial using recursion
def factorial(number):
    if number == 0 or number == 1:
        return 1

    return number * factorial(number - 1)

print(factorial(3))
```

```python
# 6. Check Armstrong number
def check_armstrong(number):
    original = number
    total = 0
    power = len(str(number))

    while number > 0:
        number,digit = divmod(number,10)
        total += digit ** power

    if total == original:
        return True
    else:
        return False

print(check_armstrong(153))
```

```python
# 7.Swap the first and last characters of a string
def swap_characters(text):
    if len(text) < 2:
        return text

    return text[::-1]

print(swap_characters("Python"))
```

```python
# 8. Convert decimal to binary
def decimal_to_binary(number):
    if number == 0:
        return "0"

    binary = ""

    while number > 0:
        number,remainder = divmod(number,2)
        binary = str(remainder) + binary

    return binary

print(decimal_to_binary(100))
```

```python
# 9. Find the longest word in a sentence
def longest_word(sentence):
    words = sentence.split()
    longest = ""

    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest

print(longest_word("Python is a powerful programming language"))
```

```python
# 10. Check whether two strings are anagrams
def check_anagram(str1, str2):
    str1 = str1.lower().replace(" ", "")
    str2 = str2.lower().replace(" ", "")

    if sorted(str1) == sorted(str2):
        return True
    else:
        return False

print(check_anagram("listen", "silent"))
```
