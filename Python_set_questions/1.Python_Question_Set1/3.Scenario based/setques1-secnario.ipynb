```python
def OddEven():
    number = int(input("Enter any number"))
    quotient, remainder = divmod(number, 2)
    if remainder == 0:
        print("Even number")
    else:
        print("Odd number")
OddEven()
```

```python
def grading_system():
    try:
        marks = int(input("Enter your marks (0-100): "))
    except ValueError:
        print("Please enter a valid number!")
        return
    if marks < 0 or marks > 100:
        print("Invalid marks")
    elif marks >= 90:
        print("Grade A")
    elif marks >= 75:
        print("Grade B")
    elif marks >= 60:
        print("Grade C")
    elif marks >= 40:
        print("Grade D - Pass")
    else:
        print("Grade F - Fail")

grading_system()
```

```python
def calculate_discount():
    total_amount = float(input("Enter your total order amount: $"))
    if total_amount > 100:
        discount = (total_amount * 20)/100
        final_amount = total_amount - discount
        print("Discount applied: $",discount)
    else:
        final_amount = total_amount
        print("No discount applied")
    print("Final amount to pay: $",final_amount)
calculate_discount()
```

```python
def bankingsystem():
    balance = float(input("Enter your current balance"))
    withdrawal = float(input("Enter your withdrawal amount"))

    if withdrawal <= 0:
        print("Invalid amount")
    elif withdrawal <= balance:
        balance = balance - withdrawal
        print("withdrawal successfully!")
        print("Ramaining balance:",balance)
    else:
        print("insufficient balance")
        print("Available balance is:",balance)
bankingsystem()
```

```python
def Leapyear():
   year = input("Enter year")
   try: 
       year = int(year)
   except ValueError:
       print("Enter a valid year")
   if year < 1000 and year > 9999:
    print("Please enter a 4-digit year")
    return
   is_leap = (year % 4 == 0 and year % 100 != 0)  or (year % 400 == 0)
   if is_leap:
       print("Leap year")
   else:
       print("Not leap year")
Leapyear()
```
